#### Feature Engineering

```python
####age feature
current_year = datetime.now().year
data['age'] = current_year - data['year']
```

```python
### milleage bucket (low/medium/high)
def mileage_bucket(x):
    if x < 50000:
        return "Low"
    elif x < 100000:
        return "Medium"
    else:
        return "High"

data['mileage_bucket'] = data['odometer'].apply(mileage_bucket)
```

```python
data.head()
```

```python
#### price per mile
data['price_per_mile'] = data['sellingprice'] / data['odometer']
data['price_per_mile'] = data['price_per_mile'].replace([float('inf')], 0)
```

```python
data.head()
```

```python
data['saledate'] = pd.to_datetime(data['saledate'], errors='coerce', utc=True,format='mixed')
data = data.dropna(subset=['saledate'])

# optional: remove timezone after standardizing
data['saledate'] = data['saledate'].dt.tz_convert(None)

data['month'] = data['saledate'].dt.month_name()
```

```python
data.head(20)
```

```python
# 6. Season Feature (CRITICAL)
def get_season(month):
    if month in ["December", "January", "February"]:
        return "Winter"
    elif month in ["March", "April", "May"]:
        return "Spring"
    elif month in ["June", "July", "August"]:
        return "Summer"
    else:
        return "Fall"
```

```python
data['season'] = data['month'].apply(get_season)
```

```python
data.tail()
```

```python
data = data.drop(columns=['year'])  # since age replaces it

```

```python
data=data.drop(['saledate',],axis=1)
data.head()
```

#### EDA (Anomaly Detection and Seasonality check)

```python
data.describe()
```

##### Anomaly Detection

```python
Q1 = data['sellingprice'].quantile(0.25)
Q3 = data['sellingprice'].quantile(0.75)
IQR = Q3 - Q1

lower_price = Q1 - 1.5 * IQR
upper_price = Q3 + 1.5 * IQR

price_outliers = data[(data['sellingprice'] < lower_price) | (data['sellingprice'] > upper_price)]

print("📌 SELLING PRICE OUTLIER RANGE")
print("Lower Bound:", lower_price)
print("Upper Bound:", upper_price)
print("Number of Outliers:", len(price_outliers))
```

```python
Q1 = data['odometer'].quantile(0.25)
Q3 = data['odometer'].quantile(0.75)
IQR = Q3 - Q1

lower_odo = Q1 - 1.5 * IQR
upper_odo = Q3 + 1.5 * IQR

odo_outliers = data[(data['odometer'] < lower_odo) | (data['odometer'] > upper_odo)]

print("📌 ODOMETER OUTLIER RANGE")
print("Lower Bound:", lower_odo)
print("Upper Bound:", upper_odo)
print("Number of Outliers:", len(odo_outliers))
```

##### Seasonality Analysis

```python
##### Monthly sales trend


monthly_sales = data['month'].value_counts()

plt.figure(figsize=(10,5))
plt.bar(monthly_sales.index, monthly_sales.values)
plt.title("Monthly Vehicle Sales Trend")
plt.xlabel("Month")
plt.ylabel("Number of Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
##### Season-wise demand
season_sales = data['season'].value_counts()

plt.figure(figsize=(6,4))
plt.bar(season_sales.index, season_sales.values)
plt.title("Season-wise Sales Distribution")
plt.xlabel("Season")
plt.ylabel("Count")
plt.show()
```

```python

##### Season vs Price

season_price = data.groupby('season')['sellingprice'].mean()

plt.figure(figsize=(6,4))
plt.bar(season_price.index, season_price.values)
plt.title("Average Price by Season")
plt.xlabel("Season")
plt.ylabel("Average Price")
plt.show()
```

```python

##### Month vs Price trend

month_price = data.groupby('month')['sellingprice'].mean()

plt.figure(figsize=(10,5))
plt.plot(month_price.index, month_price.values, marker='o')
plt.title("Average Price by Month")
plt.xlabel("Month")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.show()
```
